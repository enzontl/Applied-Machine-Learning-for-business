```python
# Vérification des imports critiques
import importlib

packages = [
    "numpy", "pandas", "scipy",
    "geopandas", "shapely", "pyproj",
    "osmnx", "networkx", "igraph",
    "folium", "matplotlib", "seaborn",
    "numba",
]

for pkg in packages:
    try:
        mod = importlib.import_module(pkg)
        version = getattr(mod, "__version__", "unknown")
        print(f"OK  {pkg:<12} {version}")
    except ImportError as e:
        print(f"KO  {pkg:<12} {e}")
```
